```python
import numpy as np
```

```python
#task1
ratings = np.array([
    [4, 5, 3, 4, 2],
    [5, 4, 4, 3, 5],
    [3, 5, 5, 4, 4],
    [4, 3, 2, 5, 4]
])

print("All Ratings:")
print(ratings)
selected_users = ratings[1:3]

print("\nRatings of 2nd and 3rd Users:")
print(selected_users)
```

```python
#task2
steps = np.array([7500, 8200, 9000, 6500, 10000,
                  7800, 8500, 9200, 7000, 11000])

high_steps = steps[steps > 8000]

print("\nDays with Steps Greater Than 8000:")
print(high_steps)
```

```python
#task3
scores = np.array([180, 195, 210, 175, 220, 190, 205, 185])

selected_scores = scores[[1, 4, 6]]  # Matches 2, 5, and 7

print("\nScores from Matches 2, 5, and 7:")
print(selected_scores)
```

```python
#task4
prices = np.array([999, 1499, 2499, 3999, 5999])

discounted_prices = prices * 0.90  # 10% discount

print("\nOriginal Prices:")
print(prices)

print("Prices After 10% Discount:")
print(discounted_prices)
```

```python
#task5
song_ratings = np.array([5, -2, 3, -1, 0, 4, -3])

song_ratings[song_ratings < 0] = 0

print("\nRatings After Replacing Negatives with 0:")
print(song_ratings)
```
