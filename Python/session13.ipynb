```python
#task1
def print_playlist_songs(songs):
    if not songs:   
        return

    print(songs[0])  
    print_playlist_songs(songs[1:])  


playlist = ["Shape Of You", "Levitating", "Senorita", "Blinding Lights"]
print_playlist_songs(playlist)
```

```python
#task2
def count_unread_messages(group):
    total = group.get('count', 0)

    for subgroup in group.get('subgroups', []):
        total += count_unread_messages(subgroup)

    return total


chat_data = {
    'count': 10,
    'subgroups': [
        {
            'count': 5,
            'subgroups': []
        },
        {
            'count': 8,
            'subgroups': [
                {
                    'count': 2,
                    'subgroups': []
                }
            ]
        }
    ]
}

print("Total unread messages:", count_unread_messages(chat_data))
```

```python
#task3
x = 'global'

def outer():
    x = 'outer'

    def inner():
        nonlocal x
        x = 'inner'

    inner()
    print('Inside outer:', x)

outer()
print('Outside:', x)
```

```python
#task4
def format_number_short(n):
    units = ['', 'K', 'M', 'B']

    def helper(num, index):
        if num < 1000 or index == len(units) - 1:
            if index == 0:
                return str(int(num))
            return f"{round(num, 1)}{units[index]}"

        return helper(num / 1000, index + 1)

    return helper(n, 0)


print(format_number_short(500))
print(format_number_short(1500))
print(format_number_short(1200000))
```
