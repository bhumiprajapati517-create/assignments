```python
#task2
steps = [4500, 7200, 9800, 10500, 12000, 8500, 11000]

i = 0

while i < len(steps):
    if steps[i] > 10000:
        print("First day you crossed 10,000 steps:", i + 1)
        break
    i += 1
```

```python
#task3
def print_team_names(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)

ipl_teams = [
    "Mumbai Indians",
    "RCB",
    "CSK",
    "Sunrisers Hyderabad",
    "Gujarat Titans"
]

print_team_names(ipl_teams)
```

```python
#task5
prices = [500, 700, 0, 400, 600, 300, 800]

total = 0

for price in prices:
    if price == 0:
        continue  

    total += price

    if total > 2000:
        print("Total exceeded ₹2000. Stopping")
        break

print("Final Total: ₹", total)
```
